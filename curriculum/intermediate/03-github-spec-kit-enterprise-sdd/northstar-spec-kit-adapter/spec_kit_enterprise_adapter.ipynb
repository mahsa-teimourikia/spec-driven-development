# Course 13 lab — Spec Kit inside an enterprise operating model

Northstar Mutual is adding AI-assisted broker-document classification. We will inspect a dated Spec Kit snapshot, bind authoritative context, enrich tasks into agent work units, test control failures, and preserve the boundary between generated proposals and organizational authority. All fixtures and evidence are synthetic.

```python
import copy
import importlib.util
import json
import sys
from pathlib import Path

lab_spec = importlib.util.spec_from_file_location('course13_notebook_lab', Path('lab.py'))
lab = importlib.util.module_from_spec(lab_spec)
sys.modules[lab_spec.name] = lab
lab_spec.loader.exec_module(lab)

source = lab.load_source()
snapshot = lab.load_snapshot()
reference = lab.load_reference()
candidate = lab.load_candidate()
print(source['context_id'], snapshot['release'], snapshot['resolved_commit'][:12])
```

## 1. Inspect the observed framework surface

A version label is not enough. The snapshot records official source, resolved commit, CLI version, command and template digests, and expected command semantics. Command presence alone is not compatibility evidence. This is reproducibility evidence—not vendor certification.

```python
print(json.dumps({
    'release': snapshot['release'],
    'commands': snapshot['core_commands'],
    'command_contracts': snapshot['command_contracts'],
    'template_digests': snapshot['template_digests'],
    'claim': snapshot['claim'],
}, indent=2))
assert lab.validate_snapshot(snapshot) == []
```

## 2. Compare safe and unsafe control packages

The reference retains source identity, open decisions, protected authority, bounded work units, and revision-bound evidence. The candidate intentionally collapses those boundaries.

```python
reference_findings = lab.review_bundle(source, snapshot, reference)
candidate_findings = lab.review_bundle(source, snapshot, candidate)
print('reference:', lab.selection_state(reference_findings), len(reference_findings))
print('candidate:', lab.selection_state(candidate_findings), len(candidate_findings))
print(json.dumps(sorted({item.code for item in candidate_findings}), indent=2))
assert reference_findings == []
assert lab.selection_state(candidate_findings) == 'BLOCKED'
```

## 3. Transform tasks into bounded agent work units

The adapter is deterministic. It carries requirement ancestry, file scope, protected decisions, stop conditions, and claim-specific evidence obligations into every consequential unit. An evidence work unit may author harnesses and manifests but cannot attest its own results.

```python
generated_units = lab.adapt_tasks_to_work_units(reference)
print(json.dumps(generated_units[0], indent=2))
assert generated_units == reference['work_units']
assert lab.validate_awu_adapter(reference) == []
```

## 4. Route by consequence, not agent preference

Risk classification belongs to trusted policy. Regulated data, AI behavior, security boundaries, and cross-repository changes select the governed route.

```python
examples = {
    'documentation': ['documentation'],
    'shared API': ['public_api'],
    'Northstar change': reference['risk_route']['signals'],
}
for name, signals in examples.items():
    print(f'{name:18} -> {lab.route_operating_model(signals)}')
route = lab.route_operating_model(reference['risk_route']['signals'])
assert route == {'governance_depth': 'consequential', 'sdd_mode': 'governed', 'execution_topology': 'multi_agent'}
```

## 5. Inject a semantic control failure

An implementation agent must not convert an open owner decision into permission. We mutate only the capability gate and confirm that the validator fails closed.

```python
mutant = copy.deepcopy(reference)
mutant['specification']['capability_gates']['automatic_requirement_satisfaction'] = 'READY'
mutant_codes = {f.code for f in lab.review_bundle(source, snapshot, mutant)}
print(sorted(mutant_codes))
assert 'BLOCKED_CAPABILITY_ENABLED' in mutant_codes
```

## 6. Test the workflow contract

Golden scenarios evaluate operational responses. Mutation cases prove that each deterministic guard can detect a deliberately introduced failure.

```python
conformance = lab.run_conformance_suite()
evaluation = lab.run_evaluation()
print(f"golden scenarios: {conformance['passed']}/{conformance['total']}")
print(f"control mutations: {evaluation['passed']}/{evaluation['total']}")
assert conformance['passed'] == conformance['total'] == 8
assert evaluation['passed'] == evaluation['total'] == 36
```

## 7. Interpret the result

`READY_FOR_OWNER_REVIEW` is intentionally weaker than approval. The framework and coding agents may propose artifacts and implementation. Trusted controls validate structure and provenance, accountable owners decide protected questions, independent systems produce evidence, and the delivery plane authorizes release. `/speckit.converge` is append-only; a legitimate semantic discovery uses a separate controlled flow-back workflow.
