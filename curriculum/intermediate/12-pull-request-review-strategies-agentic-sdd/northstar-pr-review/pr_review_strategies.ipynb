## 1. Establish the protected boundary

The agent proposes code, tests, metadata, and evidence references. The trusted application independently resolves Git state, authoritative spec and contract revisions, work-unit scope, resource-scoped reviewer authority, versioned ownership, protected CI evidence, and the terminal review decision.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
sys.path.insert(0, str(ROOT))
from review_app.simulation import (
    ci_evidence, git_backed_experiments, reference_assessment, run_reference, strategy_comparison
)
candidate, subject, decision = reference_assessment(ROOT)
assert len(candidate.base_commit) == len(candidate.head_commit) == 40
assert decision.state == 'ELIGIBLE_FOR_MERGE_REVIEW'
assert decision.merge_authorized is False
evidence = ci_evidence(subject, candidate)
assert evidence.result == 'PASS' and evidence.execution_id.endswith(subject.head_commit[:12])
(candidate.changed_paths, decision.required_scopes, decision.evidence_ids)
```

## 2. Compare strategies without inventing benchmarks

A deterministic event model runs the same synthetic change graph and review-capacity assumptions under each strategy. Keep lead time, review effort, rework, review cycles, and conflict discovery separate; these are not benchmark claims.
