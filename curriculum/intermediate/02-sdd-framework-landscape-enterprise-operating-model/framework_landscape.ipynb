# Course 12 — Framework Landscape & Enterprise Operating Model

This credential-free lab compares four **synthetic workflow styles** for Northstar AI-2310. It does not execute, benchmark, or certify named products. The safety boundary is constant: framework and agent outputs propose; trusted controls validate; accountable owners authorize.

## 1. Load one source package and two operating models

An apples-to-apples comparison starts with the same ticket, policy, domain rules, architecture, and repository discovery. The candidate is a feature contest; the reference is a hybrid operating model ready only for owner review.

```python
import copy
import importlib.util
import json
import sys
from pathlib import Path

module_spec = importlib.util.spec_from_file_location('course12_lab', Path('lab.py'))
assert module_spec and module_spec.loader
lab = importlib.util.module_from_spec(module_spec)
sys.modules[module_spec.name] = lab
module_spec.loader.exec_module(lab)

reference = lab.load_bundle()
candidate = lab.load_bundle(candidate=True)
reference_report = lab.review_operating_model(reference)
candidate_report = lab.review_operating_model(candidate)
assert lab.selection_state(reference_report) == 'READY_FOR_OWNER_REVIEW'
assert reference_report['findings'] == []
assert lab.selection_state(candidate_report) == 'OPERATING_MODEL_BLOCKED'
print('reference:', lab.selection_state(reference_report))
print('candidate:', candidate_report['counts'])
```

The reference result is intentionally narrower than adoption approval. The candidate fails because tool selection cannot repair missing authority, provenance, evidence, or execution boundaries.

## 2. Replace the feature-count baseline

A total score rewards breadth while hiding mandatory gaps. Capability profiles show what is native, extended, external, or unsupported without inventing a universal ranking.

```python
profiles = lab.capability_profiles(reference['model'])
assert profiles['overall_ranking'] is None
assert len(profiles['profiles']) == 4
for profile in profiles['profiles']:
    print(profile['id'], 'native=', len(profile['native']), 'extension=', len(profile['extension']), 'external=', len(profile['external']))
```

No style wins globally. The profile makes extension and external-control costs visible so a decision can be tied to Northstar's constraints.

## 3. Inspect transformation assurance

Specification, plan, tasks, and implementation are separate semantic transformations. Each introduces different failure modes and therefore needs different checks.

```python
complete_plan_checks = {'checks': ['requirement_coverage', 'orphan_work', 'architecture_invention', 'protected_decisions']}
assert lab.transformation_assurance('plan', complete_plan_checks) == []
unsafe_tasks = {'checks': ['scope', 'justification']}
missing = lab.transformation_assurance('tasks', unsafe_tasks)
assert missing == ['dependency', 'stop_conditions', 'work_unit_enrichment']
print('missing task assurance:', missing)
```

The task list is not safe for autonomous dispatch until dependencies, work-unit boundaries, and stop behavior are explicit.

## 4. Route proportionally by consequence

Progressive SDD uses the minimum sufficient process. Diff size alone is not a risk model.

```python
changes = [
    {'name': 'README typo'},
    {'name': 'shared API', 'shared_contract': True},
    {'name': 'regulated classifier', 'regulated_ai': True, 'sensitive_data': True},
    {'name': 'multi-repo migration', 'cross_repository': True, 'multi_agent': True},
]
routes = {item['name']: lab.route_risk_tier(item) for item in changes}
assert [routes[item['name']]['tier'] for item in changes] == [1, 2, 3, 4]
print(json.dumps(routes, indent=2))
```

The route selects an operating mode, not a global vendor. Each team can use a compatible framework configuration while preserving the same enterprise boundary.

## 5. Inject authority and portability failures

Now remove the agent boundary and make critical semantics framework-only. The validators should expose both failures independently.

```python
variant = copy.deepcopy(reference['model']['variants'][0])
variant['agent_boundary']['may_not_decide'] = []
variant['critical_semantics_survive_framework_removal'] = False
codes = {finding.code for finding in lab.validate_variant(variant, reference['model']['enterprise_requirements'])}
assert {'AGENT_BOUNDARY_MISSING', 'FRAMEWORK_LOCK_IN_HIGH'} <= codes
print(sorted(codes))
```

A human-readable file can still be governance-locked if authority or evidence only exists in framework state. Syntax portability is the weakest level.

## 6. Run migration and upgrade conformance

The migration test asks what survives removal. Golden scenarios ask whether an upgrade preserves critical semantics.

```python
for variant in reference['model']['variants']:
    result = lab.migration_survivability(variant)
    assert result['decision'] == 'PORTABLE'
conformance = lab.run_conformance_suite(reference['conformance'])
assert (conformance['passed'], conformance['population']) == (6, 6)
print(conformance['claim'])
```

The six scenarios establish only the declared synthetic adapter behavior. A production pilot still needs live framework output, representative repositories, developer experience, security, maintenance, and cost evidence.

## 7. Evaluate the deterministic rules

Thirty labelled mutations exercise source equivalence, provenance, authority, evidence, portability, selection records, extension boundaries, and upgrade assurance.

```python
evaluation = lab.evaluate_rules()
assert (evaluation['population'], evaluation['true_positive']) == (30, 30)
assert evaluation['false_positive'] == evaluation['false_negative'] == 0
assert evaluation['exact_matches'] == {'numerator': 30, 'denominator': 30}
print(json.dumps({key: evaluation[key] for key in ['population', 'true_positive', 'false_positive', 'false_negative', 'exact_matches', 'claim']}, indent=2))
```
