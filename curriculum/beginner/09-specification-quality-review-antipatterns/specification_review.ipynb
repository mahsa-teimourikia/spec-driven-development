# Course 09 lab — specification review and anti-patterns

Northstar's AI-2290 specification mentions AI, performance, security, scalability, human review, policy, and tests. We will show why concept presence is not quality, issue typed findings, inspect traceability theater, repair the package, and preserve a bounded readiness decision.

**Safety and evidence boundary:** all artifacts are fictional deterministic fixtures. The rule set exercises labelled anti-patterns; it is not a general natural-language judge, an owner approval, implementation evidence, or production-release authority.

## Objectives and architecture

You will compare an unsafe lexical baseline with field-aware review, inspect finding severity and ownership, distinguish structural from semantic traceability, test autonomy boundaries, run a labelled evaluation, and re-review a repaired package. Courses 03–08 provide hierarchy, ownership, executable requirements, evidence, and NFR contracts.

```text
spec package → review dimensions → evidence-backed findings → owner repair
                                      ↓
                           typed readiness decision
```

A reviewer proposes findings and applies an authorized gate. Product, policy, architecture, exception, and release decisions remain with their accountable owners.

````python
import copy
import importlib.util
import json
import sys
from pathlib import Path

spec = importlib.util.spec_from_file_location('course09_notebook_lab', Path('lab.py'))
assert spec and spec.loader
lab = importlib.util.module_from_spec(spec)
sys.modules[spec.name] = lab
spec.loader.exec_module(lab)

bundle = lab.load_review_bundle()
print('Course 09 fixture:', lab.run_demo()['fixture_status'])
````

## 1. Unsafe baseline — count responsible-sounding words

The baseline asks only whether seven concepts appear. This is specification theater: a presence check cannot establish meaning, population, source authority, failure behavior, evidence validity, or agent permissions.

````python
baseline = lab.lexical_presence_baseline(bundle['package'])
assert baseline['looks_complete'] is True
assert (baseline['numerator'], baseline['denominator']) == (7, 7)
print(json.dumps(baseline, indent=2))
````

The baseline is deliberately green. It is useful only as a negative example: every expected word is present while automatic mutation still lacks authority, confidence semantics, failure behavior, and bounded execution.

## 2. Review the whole package

The field-aware review inspects statements, scenario classes, policy and exception metadata, artifact structure, agent permissions, evidence, traceability, and generated context. Findings retain subject, evidence, owner, and repair.

````python
candidate_review = lab.review_bundle(bundle)
candidate_decision = lab.readiness_decision(candidate_review)
assert candidate_review['finding_counts'] == {'blocking': 16, 'review': 24, 'informational': 0}
assert len(candidate_review['finding_clusters']) == 7
assert sum(len(item['finding_ids']) for item in candidate_review['finding_clusters']) == 40
assert candidate_decision['decision'] == 'stop'
assert candidate_decision['score'] is None
print(json.dumps({'counts': candidate_review['finding_counts'], 'dimensions': candidate_review['dimensions'], 'decision': candidate_decision}, indent=2))
````

Sixteen blocking findings are not sixteen penalty points. They identify unsafe authority, autonomy, policy, scenario, evidence, exception, and context states. The gate stops affected work because at least one blocker exists; ten stylistic strengths cannot cancel it. All 40 stable machine findings remain intact while seven clusters give humans actionable remediation themes.

````python
blocking_view = [
    {key: item[key] for key in ('finding_id', 'code', 'subject_id', 'owner', 'repair')}
    for item in candidate_review['findings']
    if item['severity'] == 'blocking'
]
assert any(item['code'] == 'UNDEFINED_CONFIDENCE_SEMANTICS' for item in blocking_view)
assert any(item['code'] == 'AUTHORITY_LAUNDERING' for item in blocking_view)
print(json.dumps(blocking_view, indent=2))
````

## 3. Traceability theater

Every requirement points to `TEST-ALL`, so a link-count dashboard says 100%. Semantic review asks whether the test's declared oracle supports the exact requirement.

````python
trace = candidate_review['traceability']
assert trace['structural_link_coverage']['numerator'] == 10
assert trace['structural_link_coverage']['denominator'] == 10
assert trace['semantic_link_validity']['numerator'] == 1
assert trace['semantic_link_validity']['denominator'] == 10
print(json.dumps(trace, indent=2))
````

A structural relationship is useful for navigation but does not prove support. The same distinction applies to task, code, policy, approval, and runtime links. Course 10 will deepen the graph; here we learn to challenge the claim behind the edge. A `reviewed_by` edge proves review coverage only, never implementation or conformance.

## 4. Failure injection — false precision is not a repair

Replacing `quickly` with a number can make the requirement look stronger while preserving missing measurement and authority.

````python
precise_looking = {
    'id': 'INJECT-PERF',
    'text': 'P95 latency SHALL be below 1.7 seconds.',
    'owner': 'Commercial Underwriting Product',
    'measurement': None,
    'decision': {},
    'source_refs': [],
    'technology_names': [],
}
precision_codes = {item.code for item in lab.review_statement(precise_looking)}
assert {'FALSE_PRECISION', 'DECISION_AUTHORITY_MISSING'} <= precision_codes
print(sorted(precision_codes))
````

The repair belongs in a measurement contract and owner decision. A reviewer should not choose 1.7 seconds simply to eliminate a finding.

## 5. Failure injection — autonomy extremes

Under-constraint grants wildcard changes, no budget, no stop conditions, and deployment. Over-constraint dictates exact lines. Bounded autonomy fixes consequential decisions and leaves local implementation choices open.

````python
unsafe_codes = {item.code for item in lab.review_autonomy(bundle['autonomy'])}
reference_autonomy = lab.load_review_bundle(repaired=True)['autonomy']
assert lab.review_autonomy(reference_autonomy) == ()
over = copy.deepcopy(reference_autonomy)
over['exact_line_edits'] = [42]
over_codes = {item.code for item in lab.review_autonomy(over)}
assert 'AGENT_TASK_UNDER_CONSTRAINED' in unsafe_codes
assert 'AGENT_TASK_OVER_CONSTRAINED' in over_codes
print({'under_constrained': sorted(unsafe_codes), 'bounded': [], 'over_constrained': sorted(over_codes)})
````

## 6. Authority conflicts and self-confirming loops

Specificity, recency, and file proximity are not authority. The first extended case preserves an enterprise/feature conflict instead of synthesizing a convenient rule. The second shows one agent turning an assumption into requirement, code, tests, and its own evidence.

````python
extended_cases = {item['id']: item for item in lab.load_json(lab.EVALUATION_PATH)['cases']}
authority_codes = {item.code for item in lab.review_normative_context(extended_cases['EV-12-authority-resolution-heuristics']['input'])}
loop_codes = {item.code for item in lab.review_normative_context(extended_cases['EV-14-self-confirming-loop']['input'])}
assert {'NORMATIVE_CONFLICT', 'SPECIFICITY_MISTAKEN_FOR_AUTHORITY', 'RECENCY_MISTAKEN_FOR_AUTHORITY', 'PROXIMITY_MISTAKEN_FOR_AUTHORITY'} <= authority_codes
assert {'DERIVED_REQUIREMENT_UNAPPROVED', 'SELF_CONFIRMING_SPECIFICATION_LOOP', 'EVIDENCE_INDEPENDENCE_WEAK'} <= loop_codes
print({'authority_resolution': sorted(authority_codes), 'self_confirming_loop': sorted(loop_codes)})
````

The repair is not a smarter merge heuristic. It is independent applicability/authority resolution, preserved uncertainty, accountable approval of derived intent, and evidence outside the intent-generation loop.

## 7. Brownfield reconciliation and transition safety

Requirements, approved architecture, and repository reality can disagree independently. Review must classify matches, drift, and unknowns without treating code as truth. Desired state also needs a transition contract for mixed versions, rollback, and irreversible effects.

````python
repository_codes = {item.code for item in lab.review_brownfield_delivery(extended_cases['EV-28-repository-reconciliation']['input'])}
migration_codes = {item.code for item in lab.review_brownfield_delivery(extended_cases['EV-29-migration-gap']['input'])}
assert 'IMPLEMENTATION_VIOLATES_TRUST_BOUNDARY' in repository_codes
assert {'MIGRATION_BEHAVIOR_UNDEFINED', 'BIG_BANG_ROLLOUT_ASSUMPTION', 'ROLLBACK_SEMANTICS_UNDEFINED'} <= migration_codes
print({'repository': sorted(repository_codes), 'migration': sorted(migration_codes)})
````

A generic `spec and code differ` finding is too weak. The fixture localizes a model-facing mutation bypass and separately identifies missing discovery and legacy-pattern worship. Transition findings remain distinct from the final desired state.

## 8. Capability-scoped agent readiness

A single unresolved authorization question should not block safe extraction work, but it must not disappear into a project-level READY status. The readiness function evaluates the same explicit boundary for each capability.

````python
capability_readiness = lab.agent_readiness_by_capability(lab.load_json(lab.CAPABILITY_READINESS_PATH))
assert capability_readiness['counts'] == {'blocked': 1, 'ready_for_bounded_implementation': 4}
automatic = next(item for item in capability_readiness['capabilities'] if item['capability'] == 'automatic_mutation')
assert automatic['decision'] == 'blocked'
assert all(item['claim'] == 'implementation_readiness_only_not_production_release' for item in capability_readiness['capabilities'])
assert 'not authenticated' in capability_readiness['evidence_boundary']
print(json.dumps(capability_readiness, indent=2))
````

The output carries the ten-stage review pipeline and refuses to imply production release. Extraction, proposal creation, conflict detection, and the human-review workflow may proceed under bounded implementation while automatic mutation remains blocked. Every positive boolean is a fixture assertion linked to synthetic evidence IDs; those links teach relationship checks but are not authenticated attestations.

## 9. Labelled rule evaluation

Evaluation covers 35 positive, negative, and boundary cases across requirement quality, authority, controls, execution safety, lifecycle, brownfield migration, operability, and readiness. It reports denominators and false positives/negatives rather than claiming general reviewer intelligence.

````python
evaluation = lab.evaluate_review_rules()
assert evaluation['population'] == 35
assert evaluation['exact_case_matches'] == {'numerator': 35, 'denominator': 35}
assert evaluation['true_positive'] == 78
assert evaluation['false_positive'] == 0
assert evaluation['false_negative'] == 0
assert 'not_general' in evaluation['claim']
evaluation_summary = {key: value for key, value in evaluation.items() if key != 'cases'}
print(json.dumps(evaluation_summary, indent=2))
````

A perfect result on transparent authored cases proves only that these rules match these labels. Production review evaluation needs representative packages, independent labels, disagreement adjudication, severity calibration, drift, and false-stop/missed-blocker analysis.

## 10. Repair at the owning source and re-review

The reference package repairs requirements, policy provenance, scenarios, evidence, traceability, context integrity, artifact boundaries, and agent permissions. The review output itself does not become product or policy truth.

````python
repaired_bundle = lab.load_review_bundle(repaired=True)
repaired_review = lab.review_bundle(repaired_bundle)
repaired_decision = lab.readiness_decision(repaired_review)
assert repaired_review['findings'] == []
assert repaired_decision['decision'] == 'ready_for_bounded_implementation'
assert repaired_decision['score'] is None
assert repaired_review['traceability']['relationship_type_counts'] == {'reviewed_by': 10}
assert repaired_review['traceability']['conformance_evidence_coverage']['numerator'] == 0
print(json.dumps(repaired_decision, indent=2))
````

The decision is intentionally narrow. It authorizes no deployment, exception, or production claim. Independent implementation and release evidence remain future work.
