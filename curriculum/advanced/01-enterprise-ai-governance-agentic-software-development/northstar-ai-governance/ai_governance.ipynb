# Course 25 — Enterprise AI Governance

This notebook executes the same `governance_app` code used by the focused tests. The model or coding agent proposes; the trusted application resolves authoritative registries, derives risk, verifies evidence, authenticates approval, and enforces the protected operation.

## 1. Execute the complete reference path

The result should stop at accountable course review even though the local exact operation reaches `ALLOWED`. That local state grants no organizational or production authority.

```python
import json
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd()))
from governance_app.simulation import (LATER, make_plane, register_required, reinstatement_for, runtime_violation_for, run_reference)
ROOT = Path.cwd()
report = run_reference(ROOT)
print(json.dumps({
    'course_decision': report['decision'],
    'operation': report['reference']['operation']['decision'],
    'golden': [report['golden_scenarios']['passed'], report['golden_scenarios']['total']],
    'adversarial': [report['adversarial_evaluations']['passed'], report['adversarial_evaluations']['total']],
}, indent=2))
```

## 2. Registration and evidence are separate gates

An approved registry lifecycle does not manufacture evaluation, privacy, security, or oversight evidence. Missing evidence produces a typed uncertainty state rather than silent approval.

```python
import tempfile
with tempfile.TemporaryDirectory(prefix='course25-notebook-') as directory:
    plane = make_plane(ROOT, Path(directory) / 'governance.db')
    register_required(plane, omit={'privacy'})
    missing = plane.evaluate('AI-UC-2500', LATER)
    print(missing.decision, [item for item in missing.findings if 'PRIVACY' in item])
    plane.close()
```

## 3. A small tool change can be a major authority change

Adding an authoritative mutation tool changes the resolved subject, triggers a consequential policy floor, and conflicts with the use case's proposal-only authority. The agent cannot hide that expansion behind a confidence threshold.

```python
with tempfile.TemporaryDirectory(prefix='course25-autonomy-') as directory:
    plane = make_plane(ROOT, Path(directory) / 'governance.db')
    plane.agents['AGENT-2500']['tool_ids'].append('TOOL-UNDERWRITING-MUTATE')
    autonomy = plane.evaluate('AI-UC-2500', LATER)
    print(autonomy.decision, autonomy.risk_tier, autonomy.findings)
    plane.close()
```

## 4. Target reassessment instead of trusting or rerunning everything

The approved change map identifies which evidence obligations depend on a model or tool change. This is an impact-analysis aid, not permission to change the registry.

```python
with tempfile.TemporaryDirectory(prefix='course25-impact-') as directory:
    plane = make_plane(ROOT, Path(directory) / 'governance.db')
    model_impact = plane.change_impact('AI-UC-2500', {'model_revision': 'model-v13'})
    tool_impact = plane.change_impact('AI-UC-2500', {'agent_tools': 'expanded'})
    print(json.dumps({'model': model_impact, 'tools': tool_impact}, indent=2))
    plane.close()
```

## 5. Durable evidence and approvals survive process boundaries

The teaching database retains append-only claim evidence. Reopening the control plane preserves both the evidence history and the selected evidence-set digest used for accountable approval.

```python
with tempfile.TemporaryDirectory(prefix='course25-durable-') as directory:
    database = Path(directory) / 'governance.db'
    plane = make_plane(ROOT, database)
    register_required(plane)
    before = plane.evaluate('AI-UC-2500', LATER)
    plane.close()
    plane = make_plane(ROOT, database)
    after = plane.evaluate('AI-UC-2500', LATER)
    print(before.evidence_set_digest == after.evidence_set_digest, len(plane.evidence_history('AI-UC-2500')))
    plane.close()
```

## 6. Runtime suspension is durable; reinstatement is not automatic approval

A runtime signal is actionable only when it names a registered invariant and active deployment, has a coherent observation window and counts, and carries trusted evidence. Human review moves a suspended use case to reassessment—not directly back to operation.

```python
with tempfile.TemporaryDirectory(prefix='course25-suspension-') as directory:
    database = Path(directory) / 'governance.db'
    plane = make_plane(ROOT, database)
    observation, signature = runtime_violation_for(plane)
    plane.suspend_on_violation('AI-UC-2500', observation, signature, LATER)
    plane.close()
    plane = make_plane(ROOT, database)
    suspended = plane.lifecycle('AI-UC-2500')['lifecycle']
    remediation, review_signature = reinstatement_for(plane)
    reviewed = plane.reinstate_after_review('AI-UC-2500', remediation, review_signature, 'SESSION-AI-GOV', LATER)
    print(suspended, reviewed.decision)
    plane.close()
```
